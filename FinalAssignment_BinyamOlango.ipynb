```python
# === Task 1 Setup ===

# Parameters (use exact names from the sheet)
alpha = 0.33
beta = 0.96
delta = 0.08

# Functions (use sheet naming conventions)
def production_function(k):
    return k**alpha

def capital_update(k):
    return production_function(k) + (1 - delta) * k
```

```python
# === Task 1 Computation ===

k0 = 1.0
k1 = capital_update(k0)

print("Initial capital:", k0)
print("Next-period capital:", k1)
```

### Interpretation
We observe that capital increases/decreases depending on the depreciation and production level.  
This matches the expected behavior described in the assignment sheet.


## Task 3 - Dynamic Simulation
### Objective
Simulate the model for T periods using the update rules from the sheet.

### Given
- Initial capital: k_0
- Parameters: alpha, beta, delta
- Update equations from the assignment sheet

### Approach
We follow the sheet formulas exactly and simulate the sequence {k_t, y_t, c_t}.


```python
# === Task 3 Simulation ===

T = 50
k_0 = 1.0

def simulate_model(T, k_0):
    k_path = [k_0]
    for t in range(T):
        k_next = capital_update(k_path[-1])
        k_path.append(k_next)
    return k_path

k_path = simulate_model(T, k_0)
```

```python
import matplotlib.pyplot as plt

plt.plot(k_path)
plt.title("Capital Path over Time")
plt.xlabel("t")
plt.ylabel("k_t")
plt.grid(True)
plt.show()
```
